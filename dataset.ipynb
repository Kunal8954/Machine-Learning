# loading dataset

```python
from sklearn.datasets import load_digits, load_diabetes, load_iris
import pandas as pd
import matplotlib as plt
```

```python
data = load_diabetes()

df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

print(df)
```

```python
data = load_iris()

df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

print(df)
```

```python
data = load_digits()

df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

print(df)
```
